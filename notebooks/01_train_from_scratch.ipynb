# CIFAR-10 classifier

```python
import sys
sys.path.append('..')

from src.utils import load_cifar10, plot_examples
from src.model import ConvNet
```

## Load and inspect the data

```python
x_train, x_test, y_train, y_test = load_cifar10()
print(x_train.shape, x_test.shape)
```

```python
plot_examples(x_train, y_train)
```

## Train

```python
model = ConvNet(input_shape=(32, 32, 3), num_classes=10)
costs = model.train(
    x_train, y_train,
    epochs=30, batch_size=64, learning_rate=0.1
)
```

```python
import matplotlib.pyplot as plt

plt.plot(costs)
plt.xlabel('epoch')
plt.ylabel('cost')
plt.show()
```

## Evaluate

```python
preds = model.predict(x_test)
accuracy = (preds == y_test).mean()
print(f"Test accuracy: {accuracy:.4f}")
```
